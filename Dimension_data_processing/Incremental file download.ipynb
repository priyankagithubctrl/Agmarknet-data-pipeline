```python
#download Above MSP report CSV files into S3 bucket
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv,*/*"
}

commodity_groups = [1,2,3,4,10]
grp_name ={10:'Others',2:'Pulses',4:'Fibre_Crops',1:'Cereals',3:'Oil_Seeds'}
	
for i in commodity_groups:
    url = f'https://api.agmarknet.gov.in/v1/prices-and-arrivals/trans-above-msp?reportType=Market+Prices+Reported+Above+MSP+for+Specified+Period&dateFrom={today}&dateTo={today}&commodityGroupId={i}&includeCsv=true'
    r = requests.get(url, headers=headers, timeout=120)
    s3 = boto3.client("s3", aws_access_key_id=access_key_id, aws_secret_access_key=access_key)
    s3.put_object(
        Bucket="agmarknet-pc",
        Key=f"AgmarkIncremental/AboveMSP/{today}/AboveMSP_{grp_name[i]}_{today}.csv",
        Body=r.content
)
```
