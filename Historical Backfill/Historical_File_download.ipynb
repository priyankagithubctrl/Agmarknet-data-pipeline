```python
access_key_id = dbutils.secrets.get(scope="aws",key="access_key_id")
access_key = dbutils.secrets.get(scope="aws",key="access_key")
```

```python
#download daily arrival=- report CSV files into S3 bucket
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv,*/*"
}

api_key="579b464db66ec23bdd0000012dadcc947a8f4b916102e2348d872df8"
resource_id="35985678-0d79-46b4-9ed6-6f13308a1d24"
year=2026
for month in range(3,4):
    days = monthrange(year,month)[1]
    for date in range(1,days+1):
        url = f"https://api.data.gov.in/resource/{resource_id}?api-key={api_key}&offset=0&limit=all&format=csv&filters%5BArrival_Date%5D={date}-{month}-{year}"
        r = requests.get(url, headers=headers, timeout=120)
        s3 = boto3.client("s3", aws_access_key_id=access_key_id, aws_secret_access_key=access_key)
        s3.put_object(
            Bucket="agmarknet-pc",
            Key=f"agmark/{year}/agmarknet_daily_arrival{date}{month}{year}.csv",
            Body=r.content
)
```

```python
#download Above MSP report CSV files into S3 bucket
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv,*/*"
}

year=2026
commodity_groups = [1,2,3,4,10]
grp_name ={10:'Others',2:'Pulses',4:'Fibre_Crops',1:'Cereals',3:'Oil_Seeds'}
	
for i in commodity_groups:
    for month in range(4,7):
        dateFrom = f'01-{month:02d}-{year}'
        days = monthrange(year,month)[1]
        dateTo = f'{days:02d}-{month:02d}-{year}'
        url = f'https://api.agmarknet.gov.in/v1/prices-and-arrivals/trans-above-msp?reportType=Market+Prices+Reported+Above+MSP+for+Specified+Period&dateFrom={dateFrom}&dateTo={dateTo}&commodityGroupId={i}&includeCsv=true'
        r = requests.get(url, headers=headers, timeout=120)
        s3 = boto3.client("s3", aws_access_key_id=access_key_id, aws_secret_access_key=access_key)
        s3.put_object(
            Bucket="agmarknet-pc",
            Key=f"raw/msp/{year}/AboveMSP_{grp_name[i]}_{month}{year}.csv",
            Body=r.content
)
```
